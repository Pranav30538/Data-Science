```
#1
import pandas as pd

data = {
    'Study_Hrs': [2, 3, 5, 4, 6, 7, 8, 1, 3, 5],
    'Attendance_Pct': [80, 90, 75, 85, 95, 70, 92, 60, 88, 78],
    'Past_Score': [60, 70, 80, 75, 90, 65, 88, 55, 72, 79],
    'Result': ['Fail', 'Pass', 'Pass', 'Pass', 'Pass', 'Fail', 'Pass', 'Fail', 'Pass', 'Pass']
}
df = pd.DataFrame(data)

print("Sample DataFrame 'df' created:")
display(df.head())
```

```
#2
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import LabelEncoder

X = df[['Study_Hrs', 'Attendance_Pct', 'Past_Score']]
y = df['Result']

encoder = LabelEncoder()
y = encoder.fit_transform(y)

model = DecisionTreeClassifier(random_state=42)
model.fit(X, y)

print("Decision Tree Classifier trained successfully.")
```

```
new_student = [[4, 75, 65]]
prediction = model.predict(new_student)
print(prediction)
```

```
from sklearn.metrics import confusion_matrix, accuracy_score

y_pred = model.predict(X)

cm = confusion_matrix(y, y_pred)
accuracy = accuracy_score(y, y_pred)

print("Confusion Matrix:")
print(cm)

print("Accuracy:", accuracy * 100, "%")
```

```
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt

plt.figure(figsize=(8,6))
plot_tree(model,
          feature_names=['Study_Hrs', 'Attendance_Pct', 'Past_Score'],
          class_names=['Fail', 'Pass'],
          filled=True)

plt.show()
```

```
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(5,4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['Fail', 'Pass'],
            yticklabels=['Fail', 'Pass'])

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix Heatmap")
plt.show()
```
